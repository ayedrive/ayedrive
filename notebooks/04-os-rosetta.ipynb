# OS Rosetta Stone: boundary ownership
Classify which layer owns each responsibility. The table is a conceptual exercise; it does not claim each adapter exists.


```python
ownership = {'PnP and power': 'OS adapter', 'device wire protocol': 'portable core', 'capability dispatch': 'HAL', 'MMIO/DMA/IRQ primitives': 'backend'}
assert set(ownership.values()) == {'OS adapter', 'portable core', 'HAL', 'backend'}
for responsibility, layer in ownership.items():
    print(f'{responsibility}: {layer}')
```
