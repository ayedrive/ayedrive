# Interface impedance and four-layer boundaries
This notebook is an executable design exercise, not a driver implementation. It contrasts native OS glue, portable protocol/state logic, HAL contracts, and real/sim backends.


```python
layers = ['OS adapter', 'portable core', 'HAL contract', 'real or simulated backend']
assert len(layers) == 4
print('Layer order:', ' -> '.join(layers))
```

A profile is not a driver. Native buffers, locks, power, interrupts, and DMA ownership stay in the OS adapter/backend; wire-protocol behavior stays in the portable core only when identical across devices.
